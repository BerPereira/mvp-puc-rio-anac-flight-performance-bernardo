# 01 - Ingestão da Camada Bronze

Este notebook realiza a ingestão dos dados brutos utilizados no MVP de Engenharia de Dados.

As fontes utilizadas são dados públicos da ANAC relacionados a voos regulares e aeródromos.

Nesta etapa, os dados são preservados o mais próximo possível de sua forma original, sendo armazenados na camada Bronze para garantir rastreabilidade e permitir o reprocessamento das etapas posteriores do pipeline.

```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.mvp_sprint_3_anac;

CREATE VOLUME IF NOT EXISTS workspace.mvp_sprint_3_anac.raw_files;
```

```python
dbutils.fs.mkdirs(
    "/Volumes/workspace/mvp_sprint_3_anac/raw_files/vra"
)

dbutils.fs.mkdirs(
    "/Volumes/workspace/mvp_sprint_3_anac/raw_files/aerodromos"
)
```

```python
print("Arquivos VRA:")
for arquivo in dbutils.fs.ls(
    "/Volumes/workspace/mvp_sprint_3_anac/raw_files/vra"
):
    print(f"{arquivo.name} - {arquivo.size / 1024 / 1024:.2f} MB")

print("\nArquivo de aeródromos:")
for arquivo in dbutils.fs.ls(
    "/Volumes/workspace/mvp_sprint_3_anac/raw_files/aerodromos"
):
    print(f"{arquivo.name} - {arquivo.size / 1024 / 1024:.2f} MB")
```

```python
vra_path = "/Volumes/workspace/mvp_sprint_3_anac/raw_files/vra/*.csv"
aerodromos_path = "/Volumes/workspace/mvp_sprint_3_anac/raw_files/aerodromos/*.csv"
```

```python
from pyspark.sql.functions import col, current_timestamp

df_vra_bronze = (
    spark.read
    .option("header", True)
    .option("sep", ";")
    .option("encoding", "UTF-8")
    .csv(vra_path)
    .withColumn("arquivo_origem", col("_metadata.file_name"))
    .withColumn("data_ingestao", current_timestamp())
)

display(df_vra_bronze)
```

```python
df_vra_bronze.printSchema()
```

```python
from pyspark.sql.functions import col, current_timestamp

df_aerodromos_bronze = (
    spark.read
    .option("header", True)
    .option("sep", ",")
    .option("quote", '"')
    .option("encoding", "UTF-8")
    .csv(aerodromos_path)
    .withColumn("arquivo_origem", col("_metadata.file_name"))
    .withColumn("data_ingestao", current_timestamp())
)

display(df_aerodromos_bronze)
```

```python
df_aerodromos_bronze.printSchema()
```

```python
import re
import unicodedata

def normalizar_nome_coluna(nome):
    nome = unicodedata.normalize("NFKD", nome)
    nome = "".join(c for c in nome if not unicodedata.combining(c))
    nome = nome.lower()
    nome = re.sub(r"[^a-z0-9]+", "_", nome)
    nome = re.sub(r"_+", "_", nome).strip("_")
    return nome

def normalizar_colunas(df):
    novo_df = df

    for coluna_original in df.columns:
        coluna_nova = normalizar_nome_coluna(coluna_original)
        novo_df = novo_df.withColumnRenamed(coluna_original, coluna_nova)

    return novo_df

df_vra_bronze = normalizar_colunas(df_vra_bronze)
df_aerodromos_bronze = normalizar_colunas(df_aerodromos_bronze)

df_vra_bronze.printSchema()
```

```python
df_vra_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.mvp_sprint_3_anac.bronze_vra")
```

```python
df_aerodromos_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.mvp_sprint_3_anac.bronze_aerodromos")
```

```python
print("bronze_vra:", spark.table("workspace.mvp_sprint_3_anac.bronze_vra").count())
print("bronze_aerodromos:", spark.table("workspace.mvp_sprint_3_anac.bronze_aerodromos").count())
```

## Resultado da ingestão

A camada Bronze foi criada com sucesso no Databricks utilizando tabelas Delta.

Foram persistidas duas tabelas:

- `bronze_vra`: registros de voos regulares da ANAC referentes ao período de janeiro a julho de 2026.
- `bronze_aerodromos`: cadastro de aeródromos públicos da ANAC.

Os arquivos originais foram preservados no Volume do Unity Catalog e as tabelas Bronze mantêm os valores recebidos das fontes, adicionando apenas metadados técnicos de origem e ingestão.

Os nomes das colunas foram normalizados para o padrão `snake_case` para garantir compatibilidade com tabelas Delta, sem alteração do conteúdo semântico dos dados.
